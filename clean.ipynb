```python
import sqlite3
import pandas as pd

db = pd.read_sql_query("SELECT * FROM YEAR_2025", sqlite3.connect('database.sqlite'))
db.info()
```

```python
# Basic data validation checks
for column, count in db.drop(columns=['product', 'vendor', 'cweId']).isnull().sum().items():
    if count > 0:
        raise ValueError(f"Column '{column}' has {count} null values.")

if not db['cveId'].is_unique:
    raise ValueError("Column 'cveId' is not unique.")
```
